```python
from typing import TypedDict

# Define state outside classes
class MyState(TypedDict):
    message: str


# Define the start node
def start_node(state: MyState) -> MyState:
    print("Start node received:", state["message"])
    return {"message": state["message"] + " -> processed by start"}


# Define the end node
def end_node(state: MyState) -> MyState:
    print("End node received:", state["message"])
    return state
```

```python
#Build Graph

builder = StateGraph(MyState)
```

```python
builder.add_node("start",start_node)
builder.add_node("end",end_node)
```

```python
builder.add_edge("__start__", "start_node")     # START → start_node
builder.add_edge("start_node", "end_node")      # start → end
builder.add_edge("end_node", END)               # end → END
```

```python
graph = builder.compile()

# Run the graph
result = graph.invoke({"message": "Hello LangGraph"})

print("Final result:", result["message"])
```

```python
from typing import TypedDict
from langgraph.graph import StateGraph, END


class MyState(TypedDict):
    message: str


def start_node(state: MyState) -> MyState:
    print("Start node received:", state["message"])
    return {"message": state["message"] + " -> processed by start"}


def end_node(state: MyState) -> MyState:
    print("End node received:", state["message"])
    return state


# Build graph
builder = StateGraph(MyState)

builder.add_node("start_node", start_node)
builder.add_node("end_node", end_node)

# IMPORTANT: define entry using __start__
builder.add_edge("__start__", "start_node")     # START → start_node
builder.add_edge("start_node", "end_node")      # start → end
builder.add_edge("end_node", END)               # end → END

# Compile & run
graph = builder.compile()

result = graph.invoke({"message": "Hello LangGraph"})

print("Final result:", result["message"])
```
